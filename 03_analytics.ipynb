```python
import duckdb
import pandas as pd

df = pd.read_parquet('/Users/tejaharshitamullapudi/Downloads/brfss-clinical-ai/data/processed/brfss_clean.parquet')
print(df.shape)
```

```python
result = duckdb.query("""
    SELECT diabetes, COUNT(*) as count
    FROM df
    GROUP BY diabetes
    ORDER BY count DESC
""").df()

print(result)
```

```python
result = duckdb.query("""
    SELECT COUNT(*) as count
    FROM df
    WHERE diabetes = 'Yes' AND smoking = 'Yes'
""").df()

print(result)
```

```python
result = duckdb.query("""
    SELECT _STATE, COUNT(*) as count
    FROM df
    WHERE diabetes = 'Yes'
    GROUP BY _STATE
    ORDER BY count DESC
    LIMIT 1
""").df()

print(result)
```

```python
result = duckdb.query("""
    SELECT _STATE,
        COUNT(*) as total,
        SUM(CASE WHEN diabetes = 'Yes' THEN 1 ELSE 0 END) as diabetic_count,
        ROUND(SUM(CASE WHEN diabetes = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) as diabetes_rate
    FROM df
    GROUP BY _STATE
    ORDER BY diabetes_rate DESC
    LIMIT 5
""").df()

print(result)
```

```python
result = duckdb.query("""
    SELECT sex, ROUND(AVG(_BMI5) / 100, 2) as avg_bmi
    FROM df
    GROUP BY sex
""").df()

print(result)
```
